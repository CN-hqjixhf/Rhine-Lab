##代码

```python
scores=[58,76,92,60,45,88,100,67,60,83]
print(f"人数：{len(scores)},总分:{sum(scores)}")
print(f"平均分：{sum(scores)/len(scores):.2f}")
new_scores=sorted(scores)
print(f"最小值为：{new_scores[0]},最大值为：{new_scores[-1]}")
good=0
yes=0
nogood=0
for i in scores:
    if i>=90:
        print(f"{i}：优秀")
        good+=1
    elif i<90 and i>=60:
        print(f"{i}：及格但未优秀")
        yes+=1
    else:
        print(f"{i}：不及格，{scores.index(i)}")
        nogood+=1
print(f"优秀人数：{good},及格但未优秀人数：{yes},不及格人数：{nogood}")
def summarize_scores(score,pass_line):
    yes2=0
    for score in scores:
        if score>=pass_line:
            yes2+=1
    return yes2
if scores ==[] :
    print("ValueError")
yes3=summarize_scores(scores,60)
yes4=summarize_scores(scores,70)
dict1={}
dict1["人数:"]=len(scores)
dict1["平均分:"]=sum(scores)/len(scores)
dict1["pass_line=60时及格人数:"]=yes3
dict1["pass_line=60时及格率:"]=yes3/len(scores)
print(dict1)
dict2={}
dict2["人数:"]=len(scores)
dict2["平均分:"]=sum(scores)/len(scores)
dict2["pass_line=70时及格人数:"]=yes4
dict2["pass_line=70时及格率:"]=yes4/len(scores)
print(dict2)
```

测试输入空列表

```python
scores=[]
def summarize_scores(score,pass_line):
    yes2=0
    for score in scores:
        if score>=pass_line:
            yes2+=1
    return yes2
if scores ==[] :
    print("ValueError")
```

第一次score=60依次运行输出

```python
score=60
```

```python
print(score+5)
```

第二次score=80只运行第二单元格输出

```python
score=80
```

```python
print(score+5)
```

每次只运行一个单元格，仍然以score=60运行输出65，要变为85，则要先运行score=80再运行print(score+5)

清除输出只清除单个单元格的输出结果而不影响其他单元格，例如清除上述第二次score=80的第二个单元格，那么再运行一次输出仍为65；
重启内核会将所有单元格输出结果清除后再顺次运行一次，例如上述第二次score=80的第二个单元格输出会变为85
